## Aufgabe 2.1 - Palindrome

---
Schreiben Sie eine Funktion `is_palindrome`, die prüft, ob es sich bei einer übergebenen Zeichenkette um ein Palindrom handelt.

**Anmerkung**: Ein Palindrom ist eine Zeichenkette, die von vorn und von hinten gelesen gleich bleibt, wie die Namen `Anna` und `Otto` oder die Wörter `Reittier` und `Lagerregal`. Dabei wird die Klein- und Grossschreibung vernachläsigt.

#### Berechnen mit einer Schleife

```python
def is_palindrome(name):
    """Berechnen mit einer Schleife"""
    for i in range(len(name) // 2):
        if name[i] != name[-(i + 1)]:
            return False
    return True
```

#### Rekursive Berechnung

```python
def is_palindrome(name):
    """Rekursive Berechnung"""
    if len(name) <= 1:
        return True
    elif name[0] != name[-1]:
        return False
    else:
        return is_palindrome(name[1:-1])
```

#### Direkter Listenvergleich

```python
def is_palindrome_3(name):
    """Direkter Listenvergleich"""
    return list(name) == list(reversed(name))
```

#### Slicing und Umkehrung

```python
def is_palindrome(name):
    """Slicing und Umkehrung"""
    return name == name[::-1]
```

#### List Comprehension

```python
def is_palindrome(name):
    """List Comprehension"""
    return all(name[i] == name[-(i + 1)] for i in range(len(name) // 2))
```

## Testen der Funktion

```python
for value in ('Daniel', 'Anna', 'Peter', 'Reittier'):
    print(value, is_palindrome(value.lower()))
```
