### Dictionaries

Ein Dictionary in Python ist eine Datenstruktur, die eine Sammlung von Schlüssel-Wert-Paaren enthält. Jeder Schlüssel im Dictionary muss eindeutig sein und kann mit einem bestimmten Wert verknüpft werden. Im Python-Code wird ein Dictionary mit geschweiften Klammern {} und einem Doppelpunkt : zwischen Schlüssel und Wert erstellt.

```python
mein_dictionary = {
    "Name": "Peter",
    "Alter": 28,
    "Stadt": "Bern"
}

print(mein_dictionary)
```

### Aufgabe
Stellen Sie sich vor, wir wollen in einem Programm die Bezeichnungen für ein Datei-Menü internationalisieren, also z.B. in Deutsch, Englisch, Französisch und Italienisch anbieten.

Das Menu könnte etwa wie folgt aussehen:

fig-6.2.png

Wir wollen ein verschachteltes Dictionary verwenden, in dem wir die Übersetzung für die Menüpunkte zum Speichern, Speichern unter neuem Namen, Neue Datei öffnen, Öffnen und Drucken angeben.

Zum Übersetzen der Begriffe:

<br>

| Englisch | Deutsch         | Französisch      | Italienisch |
| :------- | :-------------- | :--------------- | :---------- |
| File     | Datei           | Fichier          | File        |
| New      | Neu             | Nouveau          | Nuovo       |
| Open     | Öffnen          | Ouvrir           | Apri        |
| Save     | Speichern       | Enregistrer      | Salva       |
| Save as  | Speichern unter | Enregistrer sous | Salva come  |

```python
# Damit wir uns nicht wiederholen und sich dirch die Wiederholung keine
# Fehler einschleichen definieren wir an dieser Stelle entsprechende konstante
# Werte für die einzelnen 'Keys' unseres 'Dictionaries'.

FILE = 'file'
NEW = 'new'
OPEN = 'open'
SAVE = 'save'
SAVE_AS = 'save as'

menu = {
    'de': {
        FILE: {
            NEW: 'Neu',
            OPEN: 'Öffnen',
            SAVE: 'Speichern',
            SAVE_AS: 'Speichern unter'
        }
    },
    'en': {
        FILE: {
            NEW: 'New',
            OPEN: 'Open',
            SAVE: 'Save',
            SAVE_AS: 'Save as'
        }
    },
    'fr': {
        FILE: {
            NEW: 'Nouveau',
            OPEN: 'Ouvrir',
            SAVE: 'Enregistrer',
            SAVE_AS: 'Enregistrer sous'
        }
    }
}

sprache = input("Bitte Sprache eingeben (de/en/fr): ")
# Plausibilisierung

print(menu[sprache][FILE][SAVE_AS])
```
