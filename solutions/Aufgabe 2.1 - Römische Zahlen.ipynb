Für diese Aufgabe benötigen wir die römischen Zahlen. Für diejenigen, die mit römischen Zahlen nicht so ganz sicher sind, sind unter [Wikipedia](https://de.wikipedia.org/wiki/R%C3%B6mische_Zahlschrift) alle notwendigen Informationen zu finden.

Schreiben Sie ein Programm, welches eine beliebige römische Zahl in eine [ganze Zahl](https://de.wikipedia.org/wiki/Ganze_Zahl) umrechnet.

### Lösungsansatz
Einlesen einer römischen Zahl $\ldots$

```python
roman_number = input('Bitte geben Sie eine römische Zahl ein: ').upper()
```

Für die Umrechnung der eingegebenen Zeichen verwenden wir die Datenstruktur eines _Dictionaries_.

```python
roman_digits = {
    'I': 1,
    'V': 5,
    'X': 10,
    'L': 50,
    'C': 100,
    'D': 500,
    'M': 1000
}
```

Die durch den Benutzer eingegebene Zahl (eingegeben als Zeichenkette) müssen wir in einem späteren Schritt in einer Schleife Zeichen für Zeichen abarbeiten und umrechnen.

Hierzu benötigen wir eine Variable, welche das jeweilige (Zwischen-) resultat abspeichert.

```python
decimal = 0
```

Wir können leider nicht immer nur Zeichen für Zeichen einlesen und die gelesenen Zeichen addieren. Es gibt auch die speziellen Konstellationen, wie z.B. die Zahl `IV`, in welcher vom Wert 5 (`V`) die Zahl 1 (`I`) subtrahiert werden muss.

Aus diesem Grund müssen das zuletzt gelesene Zeichen jeweils zwischenspeichern. Hierzu verwenden wir eine zusätzliche Variable.

```python
previous_value = 0
```

Jetzt ist alles so weit vorbereitet, sodass wir die eingegebene Zeichenkette berechnen können.

```python
for char in roman_number:
    if char in roman_digits:
        if roman_digits[char] > previous_value:
            decimal -= previous_value   # decimal = decimal - previous_value
        else:
            decimal += previous_value
    previous_value = roman_digits[char]

decimal += previous_value

print(f'Ergebnis: {decimal}.')
```

**Hinweise**
Dieser _Code_ ist leider noch nicht perfekt. Es deckt etliche Spezialfälle nicht ab.
* Die Einhabe von `VV` wird als 10 ausgegeben. Dies ist allerdings eine falsche Eingabe, da die Zahl $10$ durch `X` repräsentiert werden sollte.
