### 1.2. Übungsaufgabe
Importieren Sie nur die Funktion choice aus dem Modul random und wählen Sie einen zufälligen Wochentag aus einer Liste von Wochentagen aus.

```
from random import choice

# Liste der Wochentage
wochentage = ["Montag", "Dienstag", "Mittwoch", "Donnerstag", "Freitag", "Samstag", "Sonntag"]

# Wählt einen zufälligen Wochentag aus
zufallstag = choice(wochentage)
print(f"Zufälliger Wochentag: {zufallstag}")
```

### 1.3. Übungsaufgabe

Importieren Sie das Modul datetime mit dem Alias dt und geben Sie das aktuelle Datum und die aktuelle Uhrzeit aus.



```
import datetime as dt

# Aktuelles Datum und aktuelle Uhrzeit
jetzt = dt.datetime.now()
print(f"Aktuelles Datum und Uhrzeit: {jetzt}")

# Formatierte Ausgabe
datum_formatiert = jetzt.strftime("%d.%m.%Y")
zeit_formatiert = jetzt.strftime("%H:%M:%S")
print(f"Datum: {datum_formatiert}")
print(f"Uhrzeit: {zeit_formatiert}")
```

## 2. Externe Module in Python

Schreiben Sie ein Programm, das den aktuellen Python-Suchpfad ausgibt und zählt, wie viele Verzeichnisse enthalten sind.

```
import sys

# Ausgabe des Python-Suchpfads
print("Python-Suchpfad:")
for i, pfad in enumerate(sys.path, 1):
    print(f"{i}. {pfad}")

# Zählen der Verzeichnisse
anzahl_verzeichnisse = len(sys.path)
print(f"\nAnzahl der Verzeichnisse im Suchpfad: {anzahl_verzeichnisse}")
```

### 3.3. Übungsaufgabe

```
from tabulate import tabulate

# Beispieldaten
daten = [
    ["Max Mustermann", 1, 85.5],
    ["Erika Beispiel", 2, 92.0],
    ["Otto Normal", 3, 78.5]
]

# Tabellenkopf
kopf = ["Name", "ID", "Punktzahl"]

# Ausgabe als ASCII-Tabelle
print(tabulate(daten, headers=kopf, tablefmt="grid"))

# Ausgabe als Markdown-Tabelle
print(tabulate(daten, headers=kopf, tablefmt="pipe"))

# Ausgabe als HTML-Tabelle
print(tabulate(daten, headers=kopf, tablefmt="html"))
```

## 4. Arbeiten mit Excel-Dateien in Python

### 4.1. Übungsaufgabe

Erstellen Sie eine einfache Excel-Datei mit Namen und Noten von 5 fiktiven Studierenden. Lesen Sie diese Datei mit pandas ein und berechnen Sie den Durchschnitt der Noten.

```
import pandas as pd

# Daten für die Excel-Datei
daten = {
    'Name': ['Max Mustermann', 'Erika Beispiel', 'Otto Normal', 'Lisa Müller', 'Tim Schmidt'],
    'Matrikelnummer': [12345, 23456, 34567, 45678, 56789],
    'Note': [1.7, 2.3, 1.0, 3.7, 2.0]
}

# DataFrame erstellen
df = pd.DataFrame(daten)

# In Excel-Datei speichern
df.to_excel('studierende.xlsx', index=False)
print("Excel-Datei wurde erstellt.")
```

Dann lesen wir die Datei ein und berechnen den Durchschnitt:

```
import pandas as pd

# Excel-Datei einlesen
df = pd.read_excel('studierende.xlsx')

# Anzeigen des DataFrames
print("Eingelesene Daten:")
print(df)

# Berechnung des Durchschnitts
durchschnitt = df['Note'].mean()
print(f"\nDurchschnittliche Note: {durchschnitt:.2f}")

# Beste und schlechteste Note
beste_note = df['Note'].min()
schlechteste_note = df['Note'].max()
print(f"Beste Note: {beste_note}")
print(f"Schlechteste Note: {schlechteste_note}")
```

### 4.2. Übungsaufgabe

Filtern Sie aus dem DataFrame nur die Studierenden, die eine Note besser als 2,0 haben, und speichern Sie diese in einer neuen Excel-Datei.

```
import pandas as pd

# Excel-Datei einlesen
df = pd.read_excel('studierende.xlsx')

# Filtern der Daten (Noten besser als 2.0)
gute_noten = df[df['Note'] < 2.0]

# Anzeigen der gefilterten Daten
print("Studierende mit Noten besser als 2,0:")
print(gute_noten)

# Speichern in einer neuen Excel-Datei
gute_noten.to_excel('gute_noten.xlsx', index=False)
print("\nDie gefilterten Daten wurden in 'gute_noten.xlsx' gespeichert.")
```

Wie kann ich von einer grosse Excel-Datei nur die ersten Zeilen oder gar nur die _Header_ Informationen einlesen?

```
import pandas as pd

# Nur die ersten Zeilen einlesen (z.B. die ersten 5 Zeilen)
# nrows=5 bedeutet, dass nur 5 Zeilen eingelesen werden
kopfzeilen = pd.read_excel(excel_datei, nrows=5)

# Wenn du wirklich nur die Spaltenüberschriften (Header) brauchst
nur_header = pd.read_excel(excel_datei, nrows=0)
spalten_namen = nur_header.columns

print("Spaltennamen:", spalten_namen.tolist())
```

```
import pandas as pd

# Arbeitsblätter (Sheets) auflisten ohne die komplette Datei einzulesen
excel_file = pd.ExcelFile(excel_datei)
alle_arbeitsblätter = excel_file.sheet_names

print("Verfügbare Arbeitsblätter:", alle_arbeitsblätter)

# Jetzt nur die Kopfzeilen eines bestimmten Arbeitsblatts einlesen.
# Angenommen, wir wollen das Arbeitsblatt mit dem Namen "Tabelle1"
ausgewähltes_blatt = "Sheet2"

# Nur die Spaltenüberschriften des ausgewählten Arbeitsblatts
kopfzeilen = pd.read_excel(excel_datei,
                           sheet_name=ausgewähltes_blatt,
                           nrows=0)

spalten_namen = kopfzeilen.columns
print(f"Spaltennamen im Arbeitsblatt '{ausgewähltes_blatt}':", spalten_namen.tolist())
```

```
import pandas as pd

# Excel-Datei öffnen, ohne alle Daten einzulesen
excel_file = pd.ExcelFile(excel_datei)

# Um alle benannten Tabellen zu sehen, nutzen wir openpyxl
# (Pandas nutzt intern oft openpyxl für Excel-Operationen)
import openpyxl

wb = openpyxl.load_workbook(excel_datei, read_only=True)

# Für jedes Arbeitsblatt die benannten Tabellen auflisten
for sheet_name in wb.sheetnames:
    sheet = wb[sheet_name]
    if hasattr(sheet, 'tables'):  # Prüfen ob Tabellen existieren
        print(f"Tabellen in Arbeitsblatt '{sheet_name}':")
        for table_name in sheet.tables:
            print(f"  - {table_name}")

# Eine bestimmte Tabelle direkt einlesen.
# Die Syntax ist: excel!Tabellenname
tabellen_name = "Tabelle1"  # Der Name deiner CTRL+T Tabelle
tabellen_daten = pd.read_excel(excel_datei,
                               sheet_name=None,  # Alle Sheets durchsuchen
                               table=tabellen_name)

print(f"Kopfzeilen der Tabelle '{tabellen_name}':")
# Da wir nur die Kopfzeilen wollen, können wir nrows noch hinzufügen
tabellen_kopfzeilen = pd.read_excel(excel_datei,
                                    sheet_name=None,
                                    table=tabellen_name,
                                    nrows=0)

for sheet_name, data in tabellen_kopfzeilen.items():
    if not data.empty:  # Nur Sheets mit der gesuchten Tabelle
        print(f"Spalten in Tabelle '{tabellen_name}' (Sheet: {sheet_name}):")
        print(data.columns.tolist())
```
